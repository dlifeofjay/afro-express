# What are the Factors Influencing Top and Least paid Performing Events

```python
dfa = df0.filter(['Order ID', 'Order date', 'Buyer first name', 'Buyer last name', 'Buyer email', 'Purchaser city',
                  'Purchaser state', 'Purchaser country', 'Event name', 'Event start date', 'Event start time',
                  'Event timezone', 'Event location', 'Ticket quantity', 'Net sales', "Event ID"], axis=1)
```

# Top 5 and Least 10 Paid Performing Events

```python
df_top = dfa[dfa['Event name'].isin(dfv["Event name"].head().to_list())].reset_index(drop=True)
```

```python
df_top["Event Level"] = True
```

```python
df_least = dfa[dfa["Event name"].isin(dfv["Event name"].tail(10).to_list())].reset_index(drop=True)
```

```python
df_least["Event Level"] = False
```

```python
df_top['Days to event'] = (
    df_top['Event start date'].dt.normalize()
    - df_top['Order date'].dt.normalize()
).dt.days


df_least["Days to event"] = (
    df_least['Event start date'].dt.normalize()
    - df_least['Order date'].dt.normalize()
).dt.days
```

```python
# Top
df_top['hour'] = df_top['Order date'].dt.hour

def get_time(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'

df_top['Time of day'] = df_top['hour'].apply(get_time)

# Least
df_least['hour'] = df_least['Order date'].dt.hour

def get_time(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'

df_least['Time of day'] = df_least['hour'].apply(get_time)
```

```python
# Checking if ticket Price is an influence in quantity purchased by finding the average sales per ticket

ave_sales_top = df_top["Net sales"].sum() / df_top["Ticket quantity"].sum()
ave_sales_least = df_least["Net sales"].sum() / df_least["Ticket quantity"].sum()

labels = ["Top Events", "Least Events"]
values = [ave_sales_top, ave_sales_least]

plt.figure(figsize=(7, 5))

plt.bar(labels, values)

plt.title("Average Sales per Ticket")
plt.ylabel("Sales per Ticket")
plt.xlabel("Top and Least Performing Events")
plt.tight_layout()
plt.savefig("Aver Sales.png")
plt.show()
```

```python
new_df = pd.concat([df_top, df_least], ignore_index=True)
```

```python
# Relationship plots

new_df["Ticket price"] = new_df["Net sales"] / new_df["Ticket quantity"]

plt.figure(figsize=(12, 7))
sns.scatterplot(data=new_df, x="Ticket price", y="Ticket quantity", hue="Event Level")
plt.title("Price and Ticket Quantity relationship for High and Low Performing Events")
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.tight_layout()
plt.savefig("Price rel Tick.png")
plt.show()
```

- while the top performing events have much more earlier purchase, with the first ranging around 40 days to the event, an average person will buy ticket around a week to 2 days to the event with upto 500 counts buying in the day of the event
- The behavior for top event could be early and intensive marketing or/and an anticipating event
- Much more early ticket purchase could be fear of ticket sell out, or lesser price on early bird
- There is high activity of ticket purchase timing in the afternoon and evening timing for both least and high performing event
- Same behaviour for the last 13 days and more than 13 days in the top performing event but theres approximately 5% increase in night sales in the last 13 days
  
# Recommendations
- High price and high demand of top performing events indicates premium event, excellent marketing or a presence of value added
- Early and intense marketing can increase ticket sales significantly
- Intense marketing should be done more in the afternoon and evening time
- Attendees profile are likely to be people who have flexibility in the afternoon, evening time

```python
from NameGenderPredictor import predict_gender
import pandas as pd

# Has 87% accuracy

new_df = new_df.copy()

def predict_name_gender(first_name):
    if pd.isna(first_name) or str(first_name).strip() == "":
        return "NaN"
    try:
        prob = predict_gender(str(first_name).strip())
        if prob is None:
            return "unknown"
        if prob > 0.5:
            return "male"
        elif prob < 0.5:
            return "female"
        else:
            return "unknown"
    except Exception as e:
        return "unknown"

new_df['Gender'] = new_df['Buyer first name'].apply(predict_name_gender)


print(new_df[['Buyer first name','Buyer last name','Gender']].head(20))
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import time


df = new_df.copy()

# Initialize geocoder
geolocator = Nominatim(user_agent="purchaser_city_map")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1.2)

# Function to get lat/long
def get_lat_long(city):
    if pd.isna(city) or city == "Unknown":
        return None, None
    try:
        location = geocode(f"{city}, USA")   # Adding USA improves accuracy
        if location:
            return location.latitude, location.longitude
        else:
            return None, None
    except:
        return None, None

# Apply to your dataframe (this may take a few minutes)
print("Geocoding in progress...")

df[['Latitude', 'Longitude']] = df['Purchaser city'].apply(
    lambda x: pd.Series(get_lat_long(x))
)

print("✅ Geocoding completed!")
print(df[['Purchaser city', 'Latitude', 'Longitude']].head())
```

# Building a Machine Learning Model For Segmentation

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

cols = []
for col in clus_df.columns:
    if clus_df[col].dtype=='str':
        cols.append(col)
```

```python
cols
```

```python
for col in cols:
    clus_df[col] = le.fit_transform(clus_df[col])
    print(le.classes_)
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3, random_state=30)
model.fit(clus_df)
```

```python
labels = model.labels_
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(df_officialk.toarray() if hasattr(clus_df, "toarray") else clus_df)

plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=labels, cmap="viridis", alpha=0.6)
plt.title("KMeans Clusters (PCA Reduced)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar(label="Cluster")
plt.savefig("cluster_ae.png")
plt.show()
```

```python
centroids = model.cluster_centers_
centroids.shape, centroids
```

```python
centroids_pca = pca.transform(model.cluster_centers_)

plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=labels, cmap="viridis", alpha=0.6)

plt.scatter(centroids_pca[:,0], centroids_pca[:,1],
            c="red", marker="X", s=200, edgecolor="black", label="Centroids")

plt.title("KMeans Clusters with Centroids (PCA Reduced)", fontsize=14)
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend()
plt.savefig("cluster_ae.png")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
ss = silhouette_score(clus_df, model.labels_)
ss
```

```python
n_clusters = range(2, 13)
inertia_error = []
silhouette_scores =[]

for k in n_clusters:
    model = KMeans(n_clusters=k, random_state=30)
    model.fit(clus_df)

    inertia_error.append(model.inertia_)
    silhouette_scores.append(silhouette_score(clus_df, model.labels_))

print('inertia error type:', type(inertia_error))
print('inertia error length:', len(inertia_error))
print('inertia error:', inertia_error)

print('silhouette scores type:', type(silhouette_scores))
print('silhouette scores length:', len(silhouette_scores))
print('silhouette scores:', silhouette_scores)
```

```python
plt.figure(figsize=(5,4))
sns.lineplot(x=n_clusters, y=inertia_error)
sns.scatterplot(x=n_clusters, y=inertia_error)
plt.show
```

- Cluster 2 contains our high value attendees, They buy more tickets, pay more likely due to late fee hike or extra charges, and they purchase tickets at the night of the event
- Cluster 1 are our early bird, they tend to purchase tickets very early in lower price. This makes them a good target for prompt purchase
